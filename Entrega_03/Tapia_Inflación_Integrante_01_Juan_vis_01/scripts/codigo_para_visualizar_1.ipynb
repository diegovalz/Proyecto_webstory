```python
# Celda 1: Instalación (Asegúrate de ejecutarla)
!pip install pandas altair vl-convert-python

# Celda 2: Generación del gráfico premium
import pandas as pd
import altair as alt

# 1. CARGA DE DATOS
df = pd.read_csv('INFLACIÓN_LIMPIA.CSV')
df['fecha'] = pd.to_datetime(df['fecha'])

# 2. DISEÑO EDITORIAL (Data Journalism Style)
# Configuramos la línea de referencia histórica del 2% del Banco Central Europeo (BCE)
meta_bce = pd.DataFrame({'y': [2.0], 'label': ['Meta Histórica BCE (2%)']})

# Capa de la línea suavizada
linea = alt.Chart(df).mark_line(
    color='#c0392b', # Rojo oscuro y elegante
    strokeWidth=2.5,
    interpolate='monotone' # Interpolación que curva la línea suavemente
).encode(
    x=alt.X('fecha:T', title='', axis=alt.Axis(
        grid=False, domainColor='transparent', tickColor='transparent',
        labelColor='#7f8c8d', labelFontSize=12
    )),
    y=alt.Y('tasa_inflacion:Q', title='Tasa de Inflación (HICP) %', axis=alt.Axis(
        gridColor='#ecf0f1', gridDash=[4, 4], domainColor='transparent',
        tickColor='transparent', labelColor='#7f8c8d', labelFontSize=12
    ))
)

# Capa de área sombreada para dar profundidad visual
area = alt.Chart(df).mark_area(
    color='#c0392b',
    opacity=0.1,
    interpolate='monotone'
).encode(
    x='fecha:T',
    y='tasa_inflacion:Q'
)

# Línea punteada de referencia (La normalidad económica)
linea_meta = alt.Chart(meta_bce).mark_rule(
    color='#2c3e50', strokeDash=[5, 5], strokeWidth=1.5, opacity=0.8
).encode(y='y')

# Etiqueta de texto para la línea de referencia
texto_meta = alt.Chart(meta_bce).mark_text(
    align='left', baseline='bottom', dx=5, dy=-5,
    color='#2c3e50', fontSize=12, fontWeight='bold'
).encode(y='y', text='label')

# Puntos interactivos invisibles (Hace que el tooltip sea súper limpio al pasar el mouse)
puntos_interactivos = alt.Chart(df).mark_circle(size=100, opacity=0).encode(
    x='fecha:T',
    y='tasa_inflacion:Q',
    tooltip=[
        alt.Tooltip('fecha:T', title='Mes', format='%B %Y'),
        alt.Tooltip('tasa_inflacion:Q', title='Inflación (%)')
    ]
)

# 3. COMPOSICIÓN Y EXPORTACIÓN
grafico_premium = (area + linea_meta + texto_meta + linea + puntos_interactivos).properties(
    width=800,
    height=450,
    title=alt.TitleParams(
        text='La Falsa Excusa: Evolución de la Inflación Europea (2005-2026)',
        subtitle=[
            'A pesar del alza reciente, la economía real promedió niveles estables cercanos a la meta del BCE.',
            'La escalada de precios en la Premier League obedece a un micro-mercado ajeno a este índice.'
        ],
        fontSize=22,
        subtitleFontSize=14,
        subtitleColor='#7f8c8d',
        anchor='start',
        offset=20
    )
).configure_view(
    strokeWidth=0 # Borra el recuadro negro por defecto de Altair
).interactive(bind_y=False) # Permite desplazarse a los lados pero bloquea el zoom vertical

# Mostrar y guardar
grafico_premium.display()
grafico_premium.save('visualizacion.html')
grafico_premium.save('visualizacion.jpg', scale_factor=2.0)

print("Gráfico renderizado con diseño editorial. Archivos HTML y JPG generados.")
```
